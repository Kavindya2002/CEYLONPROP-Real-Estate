```python
Salary = Salary.dropna().copy()
```

```python
x1 = Salary.iloc[:, 0].values.reshape(-1, 1)
```

```python
x1.shape
```

```python
y = Salary.iloc[:, 4].values
```

```python
y.shape
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x1,y)
plt.xlabel("Years")
plt.ylabel("Salary")
plt.title("Years vs Salary")
plt.grid(True, linestyle="--", linewidth=0.5)
plt.show()
```

```python
x2=  Salary.iloc[:, 2].values.reshape(-1, 1)
```

```python
x2.shape
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x2,y)
plt.xlabel("Market")
plt.ylabel("Salary")
plt.title("Market vs Salary")
plt.grid(True, linestyle="--", linewidth=0.5)
plt.show()
```

```python
model_x1 = LinearRegression()
```

```python
model_x1.fit(x1, y)
```

```python
Const=sm.add_constant(x1)
```

```python
Model=sm.OLS(y,x1).fit()
Model.summary()
```
